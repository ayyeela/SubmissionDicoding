## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** Seberapa besar keterlambatan pengiriman menurunkan review score pelanggan pada pesanan Januari 2017 - Agustus 2018?
- **Pertanyaan 2:** Negara bagian mana yang mempunyai persentase pesanan terlambatnya paling tinggi dibanding dengan negara lain selama Januari 2017 - Agustus 2018
- **Pertanyaan 3:** Bagaimana tren jumlah pesanan per bulan sepanjang tahun 2017 dan bulan apa yang paling ramai?

### Pertanyaan 2: Negara bagian mana yang mempunyai persentase pesanan terlambatnya paling tinggi dibanding dengan negara lain selama Januari 2017 - Agustus 2018

#### Persentase keterlambatan per negara bagian

```python
state_late_df = all_df.groupby("customer_state").agg(
    total=("order_id", "nunique"),
    late=("is_late", "sum")
).reset_index()
state_late_df["persen_terlambat"] = state_late_df.late / state_late_df.total * 100
state_late_df = state_late_df.sort_values("persen_terlambat", ascending=False).reset_index(drop=True)

rata_nasional = all_df.is_late.mean() * 100
print("Rata-rata keseluruhan: ", round(rata_nasional, 2), "%")
state_late_df.head(10)
```

```python
colors_ = ["#72BCD4"] * 5 + ["#D3D3D3"] * (len(state_late_df) - 5)

plt.figure(figsize=(10, 7))
sns.barplot(
    x="persen_terlambat",
    y="customer_state",
    data=state_late_df,
    hue="customer_state",
    palette=dict(zip(state_late_df.customer_state, colors_)),
    legend=False
)
plt.axvline(rata_nasional, color="red", linestyle="--", label=f"Rata-rata keseluruhan ({rata_nasional:.1f}%)")
plt.title("Percentage of Late Orders by State (biru = 5 tertinggi)", loc="center", fontsize=15)
plt.ylabel(None)
plt.xlabel("% pesanan terlambat")
plt.legend()
plt.show()
```

#### Peta choropleth persentase keterlambatan

```python
import json
import urllib.request
from matplotlib.patches import Polygon
from matplotlib.collections import PatchCollection

url_geojson = "https://raw.githubusercontent.com/codeforamerica/click_that_hood/master/public/data/brazil-states.geojson"
brazil_geo = json.load(urllib.request.urlopen(url_geojson))
```

```python
persen_per_state = state_late_df.set_index("customer_state").persen_terlambat.to_dict()

patches, nilai, label = [], [], []
for fitur in brazil_geo["features"]:
    sigla = fitur["properties"]["sigla"]
    geom = fitur["geometry"]
    poligon = geom["coordinates"] if geom["type"] == "MultiPolygon" else [geom["coordinates"]]
    terbesar = max(poligon, key=lambda p: len(p[0]))
    for p in poligon:
        patches.append(Polygon(np.array(p[0]), closed=True))
        nilai.append(persen_per_state.get(sigla, np.nan))
    cincin = np.array(terbesar[0])
    label.append((sigla, cincin[:, 0].mean(), cincin[:, 1].mean()))

cmap = plt.get_cmap("YlOrRd").copy()
cmap.set_bad("#EEEEEE")

fig, ax = plt.subplots(figsize=(8, 8))
koleksi = PatchCollection(patches, cmap=cmap, edgecolor="white", linewidth=0.6)
koleksi.set_array(np.ma.masked_invalid(nilai))
ax.add_collection(koleksi)
ax.autoscale_view()
ax.set_aspect("equal")
ax.axis("off")
for sigla, x, y in label:
    v = persen_per_state.get(sigla, np.nan)
    ax.annotate(sigla, (x, y), ha="center", va="center", fontsize=8,
                color="white" if (not np.isnan(v) and v > 15) else "black")
fig.colorbar(koleksi, ax=ax, shrink=0.6, label="% pesanan terlambat")
ax.set_title("Persentase Pesanan Terlambat per Negara Bagian", fontsize=14)
plt.show()
```

### Pertanyaan 3: Bagaimana tren jumlah pesanan per bulan sepanjang tahun 2017 dan bulan apa yang paling ramai?

```python
orders_2017_df = all_df[all_df.order_purchase_timestamp.dt.year == 2017].copy()
orders_2017_df["bulan"] = orders_2017_df.order_purchase_timestamp.dt.to_period("M").astype(str)

monthly_orders_df = orders_2017_df.groupby("bulan").agg({"order_id": "nunique"}).reset_index()
monthly_orders_df.rename(columns={"order_id": "order_count"}, inplace=True)
monthly_orders_df["perubahan"] = monthly_orders_df.order_count.diff()
monthly_orders_df["pertumbuhan_persen"] = (monthly_orders_df.order_count.pct_change() * 100).round(1)
monthly_orders_df
```

```python
print("Pertumbuhan persentase terbesar:")
print(monthly_orders_df.loc[monthly_orders_df.pertumbuhan_persen.idxmax()].to_dict())
print("Penurunan persentase terbesar:")
print(monthly_orders_df.loc[monthly_orders_df.pertumbuhan_persen.idxmin()].to_dict())
print("Pertambahan jumlah terbesar:")
print(monthly_orders_df.loc[monthly_orders_df.perubahan.idxmax()].to_dict())
```

#### Gathering dan cleaning data geolokasi

```python
geolocation_df = pd.read_csv("geolocation_dataset.csv")
geolocation_df.head()
```

```python
print("Jumlah data: ", len(geolocation_df))
geolocation_df.isna().sum()
```

```python
#Menghapus titik yang invalid
geolocation_df = geolocation_df[(geolocation_df.geolocation_lat.between(-34, 6)) &
                                (geolocation_df.geolocation_lng.between(-74, -34))]
print("Jumlah data setelah dibersihkan: ", len(geolocation_df))
```

```python
#Satu Satu kode pos mempunyai banyak titik, maka diambil nilai tengahnya (median):
zip_df = geolocation_df.groupby("geolocation_zip_code_prefix").agg(
    lat=("geolocation_lat", "median"),
    lng=("geolocation_lng", "median")
).reset_index()
zip_df.head()
```

#### Menggabungkan dengan all_df

```python
geo_all_df = pd.merge(
    left=all_df,
    right=zip_df,
    how="left",
    left_on="customer_zip_code_prefix",
    right_on="geolocation_zip_code_prefix"
)
print("Pesanan tanpa koordinat: ", geo_all_df.lat.isna().sum())
geo_all_df = geo_all_df.dropna(subset=["lat", "lng"])
geo_all_df.head()
```

#### Ringkasan per negara bagian

```python
state_geo_df = geo_all_df.groupby("customer_state").agg(
    total=("order_id", "nunique"),
    late=("is_late", "sum"),
    lat=("lat", "median"),
    lng=("lng", "median")
).reset_index()
state_geo_df["persen_terlambat"] = state_geo_df.late / state_geo_df.total * 100
state_geo_df.sort_values("persen_terlambat", ascending=False).head(6)
```

#### Faktor operasional yang bisa menyebabkan keterlambatan

```python
order_items_df = pd.read_csv("order_items_dataset.csv")
sellers_df = pd.read_csv("sellers_dataset.csv")

seller_df = (order_items_df[order_items_df.order_item_id == 1][["order_id", "seller_id"]]
             .merge(sellers_df[["seller_id", "seller_zip_code_prefix", "seller_state"]], on="seller_id", how="left")
             .merge(zip_df.rename(columns={"lat": "seller_lat", "lng": "seller_lng"}),
                    left_on="seller_zip_code_prefix", right_on="geolocation_zip_code_prefix", how="left")
             [["order_id", "seller_state", "seller_lat", "seller_lng"]])

ops_df = geo_all_df.merge(seller_df, on="order_id", how="left").dropna(subset=["seller_lat", "seller_lng"]).copy()
print("Pesanan yang dianalisis: ", len(ops_df))
```

```python
def hitung_jarak_km(lat1, lng1, lat2, lng2):
    p = np.pi / 180
    a = 0.5 - np.cos((lat2 - lat1) * p) / 2 + np.cos(lat1 * p) * np.cos(lat2 * p) * (1 - np.cos((lng2 - lng1) * p)) / 2
    return 12742 * np.arcsin(np.sqrt(a))

ops_df["jarak_km"] = hitung_jarak_km(ops_df.seller_lat, ops_df.seller_lng, ops_df.lat, ops_df.lng)
ops_df["proses_seller_hari"] = (ops_df.order_delivered_carrier_date - ops_df.order_purchase_timestamp).dt.total_seconds() / 86400
ops_df["transit_kurir_hari"] = (ops_df.order_delivered_customer_date - ops_df.order_delivered_carrier_date).dt.total_seconds() / 86400
```

```python
ops_wilayah_df = ops_df.groupby("wilayah").agg(
    jumlah=("order_id", "nunique"),
    persen_terlambat=("is_late", lambda x: x.mean() * 100),
    median_jarak_km=("jarak_km", "median"),
    median_proses_seller=("proses_seller_hari", "median"),
    median_transit_kurir=("transit_kurir_hari", "median"),
    median_estimasi=("estimasi_hari", "median")
).round(1).sort_values("persen_terlambat", ascending=False).reset_index()
ops_wilayah_df
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
metrik = [("median_jarak_km", "Median jarak seller-pelanggan (km)"),
          ("median_proses_seller", "Median proses seller (hari)"),
          ("median_transit_kurir", "Median transit kurir (hari)")]
for ax, (kolom, judul) in zip(axes, metrik):
    sns.barplot(data=ops_wilayah_df, x="wilayah", y=kolom, color="#72BCD4", ax=ax)
    for container in ax.containers:
        ax.bar_label(container, fmt="%.1f")
    ax.set_title(judul)
    ax.set_xlabel(None)
    ax.set_ylabel(None)
    ax.tick_params(axis="x", rotation=20)
plt.tight_layout()
plt.show()
```
