## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** Seberapa besar keterlambatan pengiriman menurunkan review score pelanggan pada pesanan Januari 2017 - Agustus 2018?
- **Pertanyaan 2:** Negara bagian mana yang mempunyai persentase pesanan terlambatnya paling tinggi dibanding dengan negara lain selama Januari 2017 - Agustus 2018
- **Pertanyaan 3:** Bagaimana tren jumlah pesanan per bulan sepanjang tahun 2017 dan bulan apa yang paling ramai?

## Visualization & Explanatory Analysis

### Pertanyaan 1: Seberapa besar keterlambatan pengiriman menurunkan review score pelanggan pada pesanan Januari 2017 - Agustus 2018?

```python
review_status_df = all_df.dropna(subset=["review_score"]).groupby("status_kirim").review_score.mean().round(2).reset_index()
review_status_df
```

```python
plt.figure(figsize=(6, 4))
sns.barplot(
    x="status_kirim",
    y="review_score",
    data=review_status_df.sort_values(by="review_score"),
    hue="status_kirim",
    palette={"Terlambat": "#72BCD4", "Tepat waktu": "#D3D3D3"},
    legend=False
)
plt.title("Average Review Score: Late vs On-Time Delivery", loc="center", fontsize=15)
plt.ylabel(None)
plt.xlabel(None)
plt.ylim(0, 5)
plt.show()
```

### Pertanyaan 2: Negara bagian mana yang mempunyai persentase pesanan terlambatnya paling tinggi dibanding dengan negara lain selama Januari 2017 - Agustus 2018

```python
state_late_df = all_df.groupby("customer_state").agg(total=("is_late", "size"), late=("is_late", "sum"))
state_late_df["persen_terlambat"] = state_late_df.late / state_late_df.total * 100
state_late_df = state_late_df.sort_values("persen_terlambat", ascending=False).reset_index()
state_late_df.head(10)
```

```python
colors_ = ["#72BCD4"] * 5 + ["#D3D3D3"] * (len(state_late_df) - 5)

plt.figure(figsize=(10, 7))
sns.barplot(
    x="persen_terlambat",
    y="customer_state",
    data=state_late_df,
    hue="customer_state",
    palette=dict(zip(state_late_df.customer_state, colors_)),
    legend=False
)
plt.title("Percentage of Late Orders by State", loc="center", fontsize=15)
plt.ylabel(None)
plt.xlabel(None)
plt.tick_params(axis="y", labelsize=12)
plt.show()
```

### Pertanyaan 3: Bagaimana tren jumlah pesanan per bulan sepanjang tahun 2017 dan bulan apa yang paling ramai?

## Analisis Lanjutan (Opsional) Geospatial

#### Gathering dan cleaning data geolokasi

```python
geolocation_df = pd.read_csv("geolocation_dataset.csv")
geolocation_df.head()
```

```python
print("Jumlah data: ", len(geolocation_df))
geolocation_df.isna().sum()
```

```python
#Menghapus titik yang invalid
geolocation_df = geolocation_df[(geolocation_df.geolocation_lat.between(-34, 6)) &
                                (geolocation_df.geolocation_lng.between(-74, -34))]
print("Jumlah data setelah dibersihkan: ", len(geolocation_df))
```

```python
#Satu Satu kode pos mempunyai banyak titik, maka diambil nilai tengahnya (median):
zip_df = geolocation_df.groupby("geolocation_zip_code_prefix").agg(
    lat=("geolocation_lat", "median"),
    lng=("geolocation_lng", "median")
).reset_index()
zip_df.head()
```

#### Menggabungkan dengan all_df

```python
geo_all_df = pd.merge(
    left=all_df,
    right=zip_df,
    how="left",
    left_on="customer_zip_code_prefix",
    right_on="geolocation_zip_code_prefix"
)
print("Pesanan tanpa koordinat: ", geo_all_df.lat.isna().sum())
geo_all_df = geo_all_df.dropna(subset=["lat", "lng"])
geo_all_df.head()
```

#### Ringkasan per negara bagian

```python
state_geo_df = geo_all_df.groupby("customer_state").agg(
    total=("order_id", "nunique"),
    late=("is_late", "sum"),
    lat=("lat", "median"),
    lng=("lng", "median")
).reset_index()
state_geo_df["persen_terlambat"] = state_geo_df.late / state_geo_df.total * 100
state_geo_df.sort_values("persen_terlambat", ascending=False).head(6)
```

#### Pengelompokan ke 5 wilayah Brasil

```python
region_map = {
    "AC": "Utara", "AM": "Utara", "AP": "Utara", "PA": "Utara", "RO": "Utara", "RR": "Utara", "TO": "Utara",
    "AL": "Timur Laut", "BA": "Timur Laut", "CE": "Timur Laut", "MA": "Timur Laut", "PB": "Timur Laut",
    "PE": "Timur Laut", "PI": "Timur Laut", "RN": "Timur Laut", "SE": "Timur Laut",
    "DF": "Tengah-Barat", "GO": "Tengah-Barat", "MT": "Tengah-Barat", "MS": "Tengah-Barat",
    "ES": "Tenggara", "MG": "Tenggara", "RJ": "Tenggara", "SP": "Tenggara",
    "PR": "Selatan", "RS": "Selatan", "SC": "Selatan"
}
geo_all_df["wilayah"] = geo_all_df.customer_state.map(region_map)
```

```python
region_df = geo_all_df.groupby("wilayah").agg(
    total=("order_id", "nunique"),
    persen_terlambat=("is_late", lambda x: x.mean() * 100)
).sort_values("persen_terlambat", ascending=False).round(2)
region_df
```

#### Visualisasi peta sebaran keterlambatan

```python
plt.figure(figsize=(8, 8))
scatter = plt.scatter(
    state_geo_df.lng,
    state_geo_df.lat,
    s=state_geo_df.total / 20,
    c=state_geo_df.persen_terlambat,
    cmap="YlOrRd",
    edgecolor="grey",
    alpha=0.85
)
for _, row in state_geo_df.iterrows():
    plt.annotate(row.customer_state, (row.lng, row.lat), ha="center", va="center", fontsize=8)
plt.colorbar(scatter, label="% Pesanan Terlambat")
plt.title("Late Delivery Rate by State (bubble size = number of orders)", loc="center", fontsize=14)
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.show()
```

**Insight:** (Opsional)
**Insight:**
- Wilayah timur laut punya keterlambatan tertinggi (12,72%) sekitar dua kali lipat wilayah tenggara (6,12%) dan selatan (5,90%)
- Lima negara bagian dengan keterlambatan tertinggi (AL, MA, SE, PI, CE) semuanya berada di timur laut dan membentuk satu kelompok yang jelas di peta
- Wilayah tenggara memperoleh pesanan terbanyak (65.978) dengan keterlambatan rendah sehingga pusat bisnis dan logistik kemungkinan ada di sana, jauh dari wilayah dengan keterlambatan tinggi
